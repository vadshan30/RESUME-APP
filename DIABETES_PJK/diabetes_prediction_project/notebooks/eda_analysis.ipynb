# Exploratory Data Analysis (EDA) - Diabetes Risk Prediction Dataset

This notebook covers the exploratory data analysis for the Pima Indians Diabetes Dataset.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set plotting style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
COLUMNS = ['Pregnancies', 'Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI', 'DiabetesPedigreeFunction', 'Age', 'Outcome']
df = pd.read_csv('../data/diabetes.csv', names=COLUMNS)
df.head()
```

## 1. Age vs Diabetes Distribution

```python
plt.figure(figsize=(10, 6))
sns.histplot(data=df, x="Age", hue="Outcome", multiple="stack", kde=True, palette="bright")
plt.title('Age Distribution by Diabetes Outcome')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

## 2. Glucose vs Outcome

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x="Outcome", y="Glucose", data=df, palette="Set2")
plt.title('Glucose Levels by Diabetes Outcome')
plt.xlabel('Outcome (0: Non-Diabetic, 1: Diabetic)')
plt.ylabel('Glucose Level')
plt.show()
```

## 3. Feature Correlation Heatmap

```python
plt.figure(figsize=(12, 8))
correlation_matrix = df.corr()
sns.heatmap(correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5)
plt.title('Feature Correlation Heatmap')
plt.show()
```

## 4. BMI Distribution

```python
plt.figure(figsize=(10, 6))
sns.histplot(df["BMI"], bins=30, kde=True, color="purple")
plt.title('BMI Distribution')
plt.xlabel('BMI')
plt.ylabel('Frequency')
plt.show()
```

## 5. Outcome Count Plot

```python
plt.figure(figsize=(8, 5))
sns.countplot(x="Outcome", data=df, palette="pastel")
plt.title('Count of Diabetic vs Non-Diabetic Patients')
plt.xlabel('Outcome')
plt.ylabel('Count')
plt.xticks(ticks=[0, 1], labels=['Non-Diabetic (0)', 'Diabetic (1)'])
plt.show()
```
